# Portable result notebook

This notebook reads the released source data from the directory next to the
repository. Set the `WALKABILITY_SOURCE_DATA` environment variable to use a
different location. Outputs are written to `outputs/figures`.


```python
from pathlib import Path
import os

WORKING_DIR = Path.cwd().resolve()
if (WORKING_DIR / "pyproject.toml").exists():
    REPO_ROOT = WORKING_DIR
elif WORKING_DIR.name == "Result replication" and (WORKING_DIR.parent / "pyproject.toml").exists():
    REPO_ROOT = WORKING_DIR.parent
else:
    raise RuntimeError("Run this notebook from the repository root or Result replication directory")

SOURCE_DATA_DIR = Path(os.environ.get(
    "WALKABILITY_SOURCE_DATA",
    REPO_ROOT.parent / "source data",
)).resolve()
OUTPUT_DIR = (REPO_ROOT / "outputs" / "figures").resolve()
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

if not SOURCE_DATA_DIR.exists():
    raise FileNotFoundError(
        f"Source data not found at {SOURCE_DATA_DIR}. Place it next to the repository "
        "or set WALKABILITY_SOURCE_DATA to its location."
    )

print(f"Source data: {SOURCE_DATA_DIR}")
print(f"Figure output: {OUTPUT_DIR}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter
import warnings

warnings.filterwarnings("ignore")

plt.rcParams['xtick.labelsize'] = 19
plt.rcParams['ytick.labelsize'] = 19
plt.rcParams['axes.labelsize'] = 19

label_dict = {0: 'Dense', 1: 'Sparse'}
COLORS = ['#E998A2', '#66C2A5']
STYPES = ['-', '-.']
MARKERS = ['o', 's']
MARKERSIZES = [6, 6]
WIDTHS = [2, 2]

City_plot_configs = {
    "Beijing":
        [{'ax_idx': 0, 'metric': 'SWI', 'show_legend': True, 'yticks': [-2, 0, 2, 4, 6, 8]},
         {'ax_idx': 1, 'metric': 'DWI', 'show_legend': False, 'yticks': [-6, -4, -2, 0, 2]},
         {'ax_idx': 2, 'metric': 'Composite', 'show_legend': False, 'yticks': [0, -2, -4, -6]}],
    "Shijiazhuang":
        [{'ax_idx': 0, 'metric': 'SWI', 'show_legend': True, 'yticks': [-2, 0, 2, 4, 6, 8]},
         {'ax_idx': 1, 'metric': 'DWI', 'show_legend': False, 'yticks': [-8, -6, -4, -2, 0, 2]},
         {'ax_idx': 2, 'metric': 'Composite', 'show_legend': False, 'yticks': [0, -2, -4, -6]}],
    "Suzhou":
        [{'ax_idx': 0, 'metric': 'SWI', 'show_legend': True, 'yticks': [-4, -2, 0, 2, 4, 6, 8]},
         {'ax_idx': 1, 'metric': 'DWI', 'show_legend': False, 'yticks': [0, 2, 4, 6, 8, 10]},
         {'ax_idx': 2, 'metric': 'Composite', 'show_legend': False, 'yticks': [-2, -1, 0, 1, 2, 3]}],
    "Dongguan":
        [{'ax_idx': 0, 'metric': 'SWI', 'show_legend': True, 'yticks': [-2, 0, 2, 4, 6, 8]},
         {'ax_idx': 1, 'metric': 'DWI', 'show_legend': False, 'yticks': [0, 4, 8, 12]},
         {'ax_idx': 2, 'metric': 'Composite', 'show_legend': False, 'yticks': [-2, 0, 2, 4, 6]}]}

for city_name in ["Beijing", "Shijiazhuang", "Suzhou", "Dongguan"]:
    print(f"(Amenity localization) Generating plot for {city_name}...")
    data_df = pd.read_csv(
        SOURCE_DATA_DIR / 'model simulation' / f'Amenity localization ({city_name}).csv')
    fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
    plot_configs = City_plot_configs[city_name]

    for config in plot_configs:
        ax = axes[config['ax_idx']]
        box = ax.get_position()
        ax.set_position([box.x0, box.y0, box.width * 1.08, box.height])
        x = np.arange(0, 1.05, 0.1)
        y_high = data_df[f"delta_PE_{config['metric']} (dense neighborhood)"].values
        y_low = data_df[f"delta_PE_{config['metric']} (sparse neighborhood)"].values

        ax.scatter(x, y_high, c=COLORS[0], s=MARKERSIZES[0], marker=MARKERS[1])
        ax.plot(x, y_high, color=COLORS[0], linewidth=WIDTHS[0],
                linestyle=STYPES[0], marker=MARKERS[0],
                markersize=MARKERSIZES[0], label=label_dict[0])
        ax.scatter(x, y_low, c=COLORS[1], s=MARKERSIZES[1], marker=MARKERS[1])
        ax.plot(x, y_low, color=COLORS[1], linewidth=WIDTHS[1],
                linestyle=STYPES[1], marker=MARKERS[1],
                markersize=MARKERSIZES[1], label=label_dict[1])
        ax.plot([-0.05, 1.05], [0, 0], linestyle='--', color='k', linewidth=1)
        ax.set_xlim([-0.05, 1.05])
        ax.set_yticks(config['yticks'])

        ax.yaxis.set_major_formatter(PercentFormatter(decimals=0))

        ax.set_xlabel(r'$\alpha_{POI}$')
        ax.set_ylabel(f'Δ$PE_{{{config["metric"]}}}$')

        if config['show_legend']:
            ax.legend(loc='best', labelspacing=0.2, fontsize=19)

    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / f'Amenity_localization {city_name}.svg', bbox_inches='tight')
    plt.show()
    plt.close(fig)

    print(f"(Amenity localization) Successfully saved and closed plot for {city_name}.\n")
```
