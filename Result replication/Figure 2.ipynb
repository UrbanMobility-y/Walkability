```python
import pandas as pd
import numpy as np
import geopandas
import matplotlib.pyplot as plt
import statsmodels.api as sm
import warnings

warnings.filterwarnings("ignore")

for city_name in ['Beijing', 'Shijiazhuang', 'Suzhou', 'Dongguan']:
    print(f"Generating plot for {city_name}...")

    grid_shp = geopandas.read_file(
        SOURCE_DATA_DIR / 'neighborhood-level walkability' / f'{city_name}_neighborhoods.json')

    grid_shp = grid_shp.to_crs("EPSG:4793")
    df = grid_shp[(grid_shp['SWI'] >= 0) & (grid_shp['DWI'] >= 0)].copy()
    feature_x = 'SWI'
    feature_y = 'DWI'

    COLORS = ['#E998A2', '#FEE08B', '#66C2A5']
    STYPES = ['-', '--', '-.']
    WIDTHS = [3, 3, 3]
    n_bins = 1000
    max_degree = 2

    plt.rcParams['xtick.labelsize'] = 16
    plt.rcParams['ytick.labelsize'] = 16
    plt.rcParams['axes.labelsize'] = 16

    fig, ax = plt.subplots(figsize=(3.5, 3.5))
    sub = df.copy()
    x_min = np.percentile(sub[feature_x].values, 0)
    x_max = np.percentile(sub[feature_x].values, 100)
    bin_width = (x_max - x_min) / n_bins

    bins = np.arange(x_min, x_max + bin_width, bin_width)
    labels = (bins[:-1] + bins[1:]) / 2
    sub['dist_bin'] = pd.cut(sub[feature_x], bins=bins, labels=labels, include_lowest=True)

    result = (sub
              .groupby('dist_bin', observed=False)
              .agg(mean=(feature_y, 'mean'),
                   std=(feature_y, 'std'),
                   n=(feature_y, 'count'))
              .reset_index()
              ).dropna(subset=['mean'])

    result = result[result['n'] >= 2].copy()
    result['dist_bin'] = result['dist_bin'].astype(float)

    x = result['dist_bin'].to_numpy()
    y = result['mean'].to_numpy()
    n = result['n'].to_numpy()

    best_aic = np.inf
    best_degree = 1
    best_res = None

    for deg in range(1, max_degree + 1):
        X = np.column_stack([x ** i for i in range(deg + 1)])
        wls = sm.WLS(y, X, weights=n)
        res = wls.fit()
        if res.aic < best_aic:
            best_aic = res.aic
            best_degree = deg
            best_res = res

    x_smooth = np.linspace(0, 1, 200)
    X_pred = np.column_stack([x_smooth ** i for i in range(best_degree + 1)])
    pred = best_res.get_prediction(X_pred)
    dfp = pred.summary_frame(alpha=0.05)

    y_hat = dfp['mean']
    ci_low = dfp['mean_ci_lower']
    ci_up = dfp['mean_ci_upper']

    ax.fill_between(x_smooth, ci_low, ci_up, color=COLORS[2], alpha=0.15, zorder=2)

    ax.plot(x_smooth, y_hat, color=COLORS[2], linewidth=WIDTHS[2], linestyle=STYPES[2], zorder=3)
    ax.set_xlabel(r'$Neighborhood_{SWI}$', labelpad=5)
    ax.set_ylabel(r'$Neighborhood_{DWI}$', labelpad=5)
    ax.set_xlim(0, 1)
    ax.set_ylim([0, 1])
    ax.set_xticks([0, 0.25, 0.5, 0.75, 1])
    ax.set_yticks([0, 0.25, 0.5, 0.75, 1])

    ax.grid(True, linestyle='--', linewidth=0.5, alpha=0.3, color='#7f7f7f')

    plt.savefig(OUTPUT_DIR / f'{city_name}_{feature_x}_{feature_y}.svg', bbox_inches='tight')
    plt.show()

    plt.close(fig)
    print(f"Successfully saved and closed plot for {city_name}.\n")
```

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from matplotlib.colors import to_rgba
import warnings

warnings.filterwarnings("ignore")

FILE_PATH = Path(
    SOURCE_DATA_DIR / 'City_Morphology_Correlation.csv')
# OUTPUT_DIR is defined by the portable setup cell.

COLORS = {
    "Monocentric": "#0072B2",
    "Polycentric": "#D55E00",
}

FORM_STYLE = {
    "Radial Monocentric": {
        "broad": "Monocentric",
        "marker": "o",
    },
    "Concentric Monocentric": {
        "broad": "Monocentric",
        "marker": "s",
    },
    "Clustered Polycentric": {
        "broad": "Polycentric",
        "marker": "^",
    },
    "Dispersed Polycentric": {
        "broad": "Polycentric",
        "marker": "D",
    },
}

GROUP_ORDER = ["Monocentric", "Polycentric"]

REPRESENTATIVE_CITIES = {
    "Beijing": (8, 7),
    "Shijiazhuang": (9, 6),
    "Dongguan": (8, 7),
    "Suzhou": (8, 8),
}

plt.rcParams.update({
    "font.family": "sans-serif",
    "font.sans-serif": ["Arial", "DejaVu Sans"],
    "font.size": 9.5,
    "axes.labelsize": 9.5,
    "xtick.labelsize": 9.5,
    "ytick.labelsize": 9.5,
    "legend.fontsize": 9.5,
    "axes.linewidth": 0.75,
    "xtick.major.width": 0.7,
    "ytick.major.width": 0.7,
    "xtick.major.size": 3.0,
    "ytick.major.size": 3.0,
    "svg.fonttype": "none",
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})


def load_data():
    df = pd.read_csv(FILE_PATH).copy()

    required = {
        "city_en",
        "urban form",
        "corr_pop_swi",
        "corr_pop_dwi",
    }
    missing = required.difference(df.columns)
    if missing:
        raise ValueError(f"Missing required columns: {sorted(missing)}")

    unknown_forms = set(df["urban form"].dropna().unique()) - set(FORM_STYLE)
    if unknown_forms:
        raise ValueError(
            "Unexpected urban-form categories: "
            + ", ".join(sorted(unknown_forms))
        )

    df["broad_form"] = df["urban form"].map(
        {k: v["broad"] for k, v in FORM_STYLE.items()}
    )

    return df


def make_city_jitter(df, amplitude=0.065, seed=42):
    cities = sorted(df["city_en"].dropna().unique())
    rng = np.random.default_rng(seed)
    offsets = rng.uniform(-amplitude, amplitude, size=len(cities))
    return dict(zip(cities, offsets))


def common_correlation_limits(df, step=0.05, pad=0.025):
    values = np.concatenate([
        df["corr_pop_swi"].dropna().to_numpy(),
        df["corr_pop_dwi"].dropna().to_numpy(),
    ])

    lower = min(values.min(), 0) - pad
    upper = max(values.max(), 0) + pad

    lower = np.floor(lower / step) * step
    upper = np.ceil(upper / step) * step

    return lower, upper


def correlation_plane_limits(series, step=0.05, pad_fraction=0.04):
    values = series.dropna().to_numpy()
    lower = min(values.min(), 0)
    upper = max(values.max(), 0)

    span = max(upper - lower, step)
    lower -= span * pad_fraction
    upper += span * pad_fraction

    lower = np.floor(lower / step) * step
    upper = np.ceil(upper / step) * step

    return lower, upper


def style_axes(ax, grid_y=False):
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_color("#3A3A3A")
    ax.spines["bottom"].set_color("#3A3A3A")

    ax.tick_params(
        axis="both",
        which="major",
        direction="out",
        color="#3A3A3A",
        labelcolor="#222222",
    )

    ax.set_axisbelow(True)

    if grid_y:
        ax.grid(
            axis="y",
            color="#E6E6E6",
            linewidth=0.55,
            linestyle="-",
            zorder=0,
        )


def city_level_density_walkability_correlations(ax, df):
    legend_handles = []
    legend_labels = []

    for form, style in FORM_STYLE.items():
        subset = df.loc[df["urban form"] == form]
        broad = style["broad"]

        handle = ax.scatter(
            subset["corr_pop_swi"],
            subset["corr_pop_dwi"],
            s=40,
            marker=style["marker"],
            facecolor=COLORS[broad],
            edgecolor="white",
            linewidth=0.55,
            alpha=0.84,
            zorder=2,
        )

        legend_handles.append(handle)
        legend_labels.append(f"{form} (n={len(subset)})")

    ax.axvline(
        0,
        color="#8A8A8A",
        linewidth=0.75,
        linestyle=(0, (3, 3)),
        zorder=0,
    )
    ax.axhline(
        0,
        color="#8A8A8A",
        linewidth=0.75,
        linestyle=(0, (3, 3)),
        zorder=0,
    )

    for city in REPRESENTATIVE_CITIES:
        row = df.loc[df["city_en"] == city]
        if len(row) != 1:
            continue

        row = row.iloc[0]
        x = row["corr_pop_swi"]
        y = row["corr_pop_dwi"]
        form = row["urban form"]
        style = FORM_STYLE[form]

        ax.scatter(
            x,
            y,
            s=64,
            marker=style["marker"],
            facecolor=COLORS[style["broad"]],
            edgecolor="#202020",
            linewidth=1.0,
            alpha=1.0,
            zorder=4,
        )

    ax.set_xlabel(r"$\rho(\mathit{pop}, \mathit{Neighborhood}_{\mathit{SWI}})$")
    ax.set_ylabel(r"$\rho(\mathit{pop}, \mathit{Neighborhood}_{\mathit{DWI}})$")

    ax.set_xlim(*correlation_plane_limits(df["corr_pop_swi"]))
    ax.set_ylim(*correlation_plane_limits(df["corr_pop_dwi"]))

    ax.xaxis.set_major_locator(mticker.MultipleLocator(0.1))
    ax.yaxis.set_major_locator(mticker.MultipleLocator(0.1))

    style_axes(ax, grid_y=False)

    ax.legend(
        legend_handles,
        legend_labels,
        loc="upper center",
        bbox_to_anchor=(0.5, -0.22),
        ncol=2,
        frameon=False,
        handletextpad=0.5,
        columnspacing=1.1,
        borderaxespad=0,
    )


def draw_boxplot_above_scatter(ax, values, position, colour):
    bp = ax.boxplot(
        [values],
        positions=[position],
        widths=0.42,
        patch_artist=True,
        showfliers=False,
        whis=1.5,
        boxprops={
            "facecolor": to_rgba(colour, 0.10),
            "edgecolor": colour,
            "linewidth": 1.2,
        },
        whiskerprops={
            "color": colour,
            "linewidth": 1.0,
        },
        capprops={
            "color": colour,
            "linewidth": 1.0,
        },
        medianprops={
            "color": "#222222",
            "linewidth": 1.5,
        },
    )

    for box in bp["boxes"]:
        box.set_zorder(4)
    for whisker in bp["whiskers"]:
        whisker.set_zorder(4)
    for cap in bp["caps"]:
        cap.set_zorder(4)
    for median in bp["medians"]:
        median.set_zorder(5)


def association_by_urban_form(
        ax,
        df,
        variable,
        ylabel,
        jitter_map,
        common_ylim,
):
    for position, group in enumerate(GROUP_ORDER, start=1):
        subset = df.loc[
            df["broad_form"] == group,
            ["city_en", variable],
        ].dropna()

        values = subset[variable].to_numpy()
        colour = COLORS[group]
        jitter = subset["city_en"].map(jitter_map).to_numpy()

        ax.scatter(
            np.full(len(subset), position) + jitter,
            values,
            s=24,
            marker="o",
            facecolor=colour,
            edgecolor="white",
            linewidth=0.4,
            alpha=0.72,
            zorder=2,
        )

        draw_boxplot_above_scatter(ax, values, position, colour)

    ax.axhline(
        0,
        color="#6F6F6F",
        linewidth=1.35,
        linestyle=(0, (4, 3)),
        zorder=1,
    )

    group_counts = (
        df.groupby("broad_form")["city_en"]
        .count()
        .reindex(GROUP_ORDER)
    )

    ax.set_xticks(
        [1, 2],
        [
            f"Monocentric\n(n={group_counts['Monocentric']})",
            f"Polycentric\n(n={group_counts['Polycentric']})",
        ],
    )

    ax.set_xlim(0.58, 2.42)
    ax.set_ylim(*common_ylim)
    ax.yaxis.set_major_locator(mticker.MultipleLocator(0.2))
    ax.set_ylabel(ylabel)

    style_axes(ax, grid_y=True)


def make_figure(df):
    jitter_map = make_city_jitter(df)
    common_ylim = common_correlation_limits(df)

    fig = plt.figure(figsize=(10.8, 3.55))
    gs = fig.add_gridspec(
        1,
        3,
        width_ratios=[1.28, 1.0, 1.0],
        left=0.075,
        right=0.99,
        top=0.96,
        bottom=0.28,
        wspace=0.34,
    )

    ax_a = fig.add_subplot(gs[0, 0])
    ax_b = fig.add_subplot(gs[0, 1])
    ax_c = fig.add_subplot(gs[0, 2], sharey=ax_b)

    city_level_density_walkability_correlations(ax_a, df)

    association_by_urban_form(
        ax=ax_b,
        df=df,
        variable="corr_pop_swi",
        ylabel=r"$\rho(\mathit{pop}, \mathit{Neighborhood}_{\mathit{SWI}})$",
        jitter_map=jitter_map,
        common_ylim=common_ylim,
    )

    association_by_urban_form(
        ax=ax_c,
        df=df,
        variable="corr_pop_dwi",
        ylabel=r"$\rho(\mathit{pop}, \mathit{Neighborhood}_{\mathit{DWI}})$",
        jitter_map=jitter_map,
        common_ylim=common_ylim,
    )

    ax_c.tick_params(axis="y", labelleft=True)

    return fig


print("Generating combined city morphology analysis figure...")
data_df = load_data()
fig = make_figure(data_df)

fig.savefig(OUTPUT_DIR / 'combined_city_morphology_analysis.svg',
            bbox_inches="tight",
            facecolor="white",
            )
plt.show()
print("Successfully saved and closed plot.")
```

```python
import pandas as pd
import numpy as np
import geopandas
import matplotlib.pyplot as plt
import statsmodels.api as sm
import warnings

warnings.filterwarnings("ignore")

for city_name in ['Beijing', 'Shijiazhuang', 'Suzhou', 'Dongguan']:
    print(f"Generating plot for {city_name}...")

    grid_shp = geopandas.read_file(
        rf'E:\experienced walkability across urban forms\Code and Data\source data\neighborhood-level walkability\{city_name}_neighborhoods.json')

    grid_shp = grid_shp.to_crs("EPSG:4793")
    df = grid_shp[(grid_shp['SWI'] >= 0) & (grid_shp['DWI'] >= 0)].copy()
    feature_x = 'SWI'
    feature_y = 'DWI'

    COLORS = ['#E998A2', '#FEE08B', '#66C2A5']
    STYPES = ['-', '--', '-.']
    WIDTHS = [3, 3, 3]
    n_bins = 1000
    max_degree = 2

    plt.rcParams['xtick.labelsize'] = 16
    plt.rcParams['ytick.labelsize'] = 16
    plt.rcParams['axes.labelsize'] = 16

    fig, ax = plt.subplots(figsize=(3.5, 3.5))
    sub = df.copy()
    x_min = np.percentile(sub[feature_x].values, 0)
    x_max = np.percentile(sub[feature_x].values, 100)
    bin_width = (x_max - x_min) / n_bins

    bins = np.arange(x_min, x_max + bin_width, bin_width)
    labels = (bins[:-1] + bins[1:]) / 2
    sub['dist_bin'] = pd.cut(sub[feature_x], bins=bins, labels=labels, include_lowest=True)

    result = (sub
              .groupby('dist_bin', observed=False)
              .agg(mean=(feature_y, 'mean'),
                   std=(feature_y, 'std'),
                   n=(feature_y, 'count'))
              .reset_index()
              ).dropna(subset=['mean'])

    result = result[result['n'] >= 2].copy()
    result['dist_bin'] = result['dist_bin'].astype(float)

    x = result['dist_bin'].to_numpy()
    y = result['mean'].to_numpy()
    n = result['n'].to_numpy()

    best_aic = np.inf
    best_degree = 1
    best_res = None

    for deg in range(1, max_degree + 1):
        X = np.column_stack([x ** i for i in range(deg + 1)])
        wls = sm.WLS(y, X, weights=n)
        res = wls.fit()
        if res.aic < best_aic:
            best_aic = res.aic
            best_degree = deg
            best_res = res

    x_smooth = np.linspace(0, 1, 200)
    X_pred = np.column_stack([x_smooth ** i for i in range(best_degree + 1)])
    pred = best_res.get_prediction(X_pred)
    dfp = pred.summary_frame(alpha=0.05)

    y_hat = dfp['mean']
    ci_low = dfp['mean_ci_lower']
    ci_up = dfp['mean_ci_upper']

    ax.fill_between(x_smooth, ci_low, ci_up, color=COLORS[2], alpha=0.15, zorder=2)

    ax.plot(x_smooth, y_hat, color=COLORS[2], linewidth=WIDTHS[2], linestyle=STYPES[2], zorder=3)
    ax.set_xlabel(r'$Neighborhood_{SWI}$', labelpad=5)
    ax.set_ylabel(r'$Neighborhood_{DWI}$', labelpad=5)
    ax.set_xlim(0, 1)
    ax.set_ylim([0, 1])
    ax.set_xticks([0, 0.25, 0.5, 0.75, 1])
    ax.set_yticks([0, 0.25, 0.5, 0.75, 1])

    ax.grid(True, linestyle='--', linewidth=0.5, alpha=0.3, color='#7f7f7f')

    plt.savefig(f'F:/{city_name}_{feature_x}_{feature_y}.svg', bbox_inches='tight')
    plt.show()

    plt.close(fig)
    print(f"Successfully saved and closed plot for {city_name}.\n")
```

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from matplotlib.colors import to_rgba
import warnings

warnings.filterwarnings("ignore")

FILE_PATH = Path(
    rf"E:\experienced walkability across urban forms\Code and Data\source data\City_Morphology_Correlation.csv")
OUTPUT_DIR = Path("F:/")

COLORS = {
    "Monocentric": "#0072B2",
    "Polycentric": "#D55E00",
}

FORM_STYLE = {
    "Radial Monocentric": {
        "broad": "Monocentric",
        "marker": "o",
    },
    "Concentric Monocentric": {
        "broad": "Monocentric",
        "marker": "s",
    },
    "Clustered Polycentric": {
        "broad": "Polycentric",
        "marker": "^",
    },
    "Dispersed Polycentric": {
        "broad": "Polycentric",
        "marker": "D",
    },
}

GROUP_ORDER = ["Monocentric", "Polycentric"]

REPRESENTATIVE_CITIES = {
    "Beijing": (8, 7),
    "Shijiazhuang": (9, 6),
    "Dongguan": (8, 7),
    "Suzhou": (8, 8),
}

plt.rcParams.update({
    "font.family": "sans-serif",
    "font.sans-serif": ["Arial", "DejaVu Sans"],
    "font.size": 9.5,
    "axes.labelsize": 9.5,
    "xtick.labelsize": 9.5,
    "ytick.labelsize": 9.5,
    "legend.fontsize": 9.5,
    "axes.linewidth": 0.75,
    "xtick.major.width": 0.7,
    "ytick.major.width": 0.7,
    "xtick.major.size": 3.0,
    "ytick.major.size": 3.0,
    "svg.fonttype": "none",
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})


def load_data():
    df = pd.read_csv(FILE_PATH).copy()

    required = {
        "city_en",
        "urban form",
        "corr_pop_swi",
        "corr_pop_dwi",
    }
    missing = required.difference(df.columns)
    if missing:
        raise ValueError(f"Missing required columns: {sorted(missing)}")

    unknown_forms = set(df["urban form"].dropna().unique()) - set(FORM_STYLE)
    if unknown_forms:
        raise ValueError(
            "Unexpected urban-form categories: "
            + ", ".join(sorted(unknown_forms))
        )

    df["broad_form"] = df["urban form"].map(
        {k: v["broad"] for k, v in FORM_STYLE.items()}
    )

    return df


def make_city_jitter(df, amplitude=0.065, seed=42):
    cities = sorted(df["city_en"].dropna().unique())
    rng = np.random.default_rng(seed)
    offsets = rng.uniform(-amplitude, amplitude, size=len(cities))
    return dict(zip(cities, offsets))


def common_correlation_limits(df, step=0.05, pad=0.025):
    values = np.concatenate([
        df["corr_pop_swi"].dropna().to_numpy(),
        df["corr_pop_dwi"].dropna().to_numpy(),
    ])

    lower = min(values.min(), 0) - pad
    upper = max(values.max(), 0) + pad

    lower = np.floor(lower / step) * step
    upper = np.ceil(upper / step) * step

    return lower, upper


def correlation_plane_limits(series, step=0.05, pad_fraction=0.04):
    values = series.dropna().to_numpy()
    lower = min(values.min(), 0)
    upper = max(values.max(), 0)

    span = max(upper - lower, step)
    lower -= span * pad_fraction
    upper += span * pad_fraction

    lower = np.floor(lower / step) * step
    upper = np.ceil(upper / step) * step

    return lower, upper


def style_axes(ax, grid_y=False):
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_color("#3A3A3A")
    ax.spines["bottom"].set_color("#3A3A3A")

    ax.tick_params(
        axis="both",
        which="major",
        direction="out",
        color="#3A3A3A",
        labelcolor="#222222",
    )

    ax.set_axisbelow(True)

    if grid_y:
        ax.grid(
            axis="y",
            color="#E6E6E6",
            linewidth=0.55,
            linestyle="-",
            zorder=0,
        )


def city_level_density_walkability_correlations(ax, df):
    legend_handles = []
    legend_labels = []

    for form, style in FORM_STYLE.items():
        subset = df.loc[df["urban form"] == form]
        broad = style["broad"]

        handle = ax.scatter(
            subset["corr_pop_swi"],
            subset["corr_pop_dwi"],
            s=40,
            marker=style["marker"],
            facecolor=COLORS[broad],
            edgecolor="white",
            linewidth=0.55,
            alpha=0.84,
            zorder=2,
        )

        legend_handles.append(handle)
        legend_labels.append(f"{form} (n={len(subset)})")

    ax.axvline(
        0,
        color="#8A8A8A",
        linewidth=0.75,
        linestyle=(0, (3, 3)),
        zorder=0,
    )
    ax.axhline(
        0,
        color="#8A8A8A",
        linewidth=0.75,
        linestyle=(0, (3, 3)),
        zorder=0,
    )

    for city in REPRESENTATIVE_CITIES:
        row = df.loc[df["city_en"] == city]
        if len(row) != 1:
            continue

        row = row.iloc[0]
        x = row["corr_pop_swi"]
        y = row["corr_pop_dwi"]
        form = row["urban form"]
        style = FORM_STYLE[form]

        ax.scatter(
            x,
            y,
            s=64,
            marker=style["marker"],
            facecolor=COLORS[style["broad"]],
            edgecolor="#202020",
            linewidth=1.0,
            alpha=1.0,
            zorder=4,
        )

    ax.set_xlabel(r"$\rho(\mathit{pop}, \mathit{Neighborhood}_{\mathit{SWI}})$")
    ax.set_ylabel(r"$\rho(\mathit{pop}, \mathit{Neighborhood}_{\mathit{DWI}})$")

    ax.set_xlim(*correlation_plane_limits(df["corr_pop_swi"]))
    ax.set_ylim(*correlation_plane_limits(df["corr_pop_dwi"]))

    ax.xaxis.set_major_locator(mticker.MultipleLocator(0.1))
    ax.yaxis.set_major_locator(mticker.MultipleLocator(0.1))

    style_axes(ax, grid_y=False)

    ax.legend(
        legend_handles,
        legend_labels,
        loc="upper center",
        bbox_to_anchor=(0.5, -0.22),
        ncol=2,
        frameon=False,
        handletextpad=0.5,
        columnspacing=1.1,
        borderaxespad=0,
    )


def draw_boxplot_above_scatter(ax, values, position, colour):
    bp = ax.boxplot(
        [values],
        positions=[position],
        widths=0.42,
        patch_artist=True,
        showfliers=False,
        whis=1.5,
        boxprops={
            "facecolor": to_rgba(colour, 0.10),
            "edgecolor": colour,
            "linewidth": 1.2,
        },
        whiskerprops={
            "color": colour,
            "linewidth": 1.0,
        },
        capprops={
            "color": colour,
            "linewidth": 1.0,
        },
        medianprops={
            "color": "#222222",
            "linewidth": 1.5,
        },
    )

    for box in bp["boxes"]:
        box.set_zorder(4)
    for whisker in bp["whiskers"]:
        whisker.set_zorder(4)
    for cap in bp["caps"]:
        cap.set_zorder(4)
    for median in bp["medians"]:
        median.set_zorder(5)


def association_by_urban_form(
        ax,
        df,
        variable,
        ylabel,
        jitter_map,
        common_ylim,
):
    for position, group in enumerate(GROUP_ORDER, start=1):
        subset = df.loc[
            df["broad_form"] == group,
            ["city_en", variable],
        ].dropna()

        values = subset[variable].to_numpy()
        colour = COLORS[group]
        jitter = subset["city_en"].map(jitter_map).to_numpy()

        ax.scatter(
            np.full(len(subset), position) + jitter,
            values,
            s=24,
            marker="o",
            facecolor=colour,
            edgecolor="white",
            linewidth=0.4,
            alpha=0.72,
            zorder=2,
        )

        draw_boxplot_above_scatter(ax, values, position, colour)

    ax.axhline(
        0,
        color="#6F6F6F",
        linewidth=1.35,
        linestyle=(0, (4, 3)),
        zorder=1,
    )

    group_counts = (
        df.groupby("broad_form")["city_en"]
        .count()
        .reindex(GROUP_ORDER)
    )

    ax.set_xticks(
        [1, 2],
        [
            f"Monocentric\n(n={group_counts['Monocentric']})",
            f"Polycentric\n(n={group_counts['Polycentric']})",
        ],
    )

    ax.set_xlim(0.58, 2.42)
    ax.set_ylim(*common_ylim)
    ax.yaxis.set_major_locator(mticker.MultipleLocator(0.2))
    ax.set_ylabel(ylabel)

    style_axes(ax, grid_y=True)


def make_figure(df):
    jitter_map = make_city_jitter(df)
    common_ylim = common_correlation_limits(df)

    fig = plt.figure(figsize=(10.8, 3.55))
    gs = fig.add_gridspec(
        1,
        3,
        width_ratios=[1.28, 1.0, 1.0],
        left=0.075,
        right=0.99,
        top=0.96,
        bottom=0.28,
        wspace=0.34,
    )

    ax_a = fig.add_subplot(gs[0, 0])
    ax_b = fig.add_subplot(gs[0, 1])
    ax_c = fig.add_subplot(gs[0, 2], sharey=ax_b)

    city_level_density_walkability_correlations(ax_a, df)

    association_by_urban_form(
        ax=ax_b,
        df=df,
        variable="corr_pop_swi",
        ylabel=r"$\rho(\mathit{pop}, \mathit{Neighborhood}_{\mathit{SWI}})$",
        jitter_map=jitter_map,
        common_ylim=common_ylim,
    )

    association_by_urban_form(
        ax=ax_c,
        df=df,
        variable="corr_pop_dwi",
        ylabel=r"$\rho(\mathit{pop}, \mathit{Neighborhood}_{\mathit{DWI}})$",
        jitter_map=jitter_map,
        common_ylim=common_ylim,
    )

    ax_c.tick_params(axis="y", labelleft=True)

    return fig


print("Generating combined city morphology analysis figure...")
data_df = load_data()
fig = make_figure(data_df)

fig.savefig("F:/combined_city_morphology_analysis.svg",
            bbox_inches="tight",
            facecolor="white",
            )
plt.show()
print("Successfully saved and closed plot.")
```
